##### 1.6. Обоснование комплексного состава выборки

Использование только новостных порталов не позволило бы получить достаточно полное представление о металлургической промышленности.

Информационные издания преимущественно отражают рыночную конъюнктуру, текущие события, прогнозы и экспертные оценки. Официальные сайты металлургических предприятий содержат первичные сведения о производственных процессах, технологиях, продукции и результатах деятельности. Научные организации обеспечивают дополнительное представление о технологических и материаловедческих аспектах отрасли.

Таким образом, сформированная выборка объединяет:

- вторичные источники — новостные и аналитические публикации;
- первичные источники — официальные сведения производителей;
- институциональные источники — отраслевые ассоциации и профессиональные объединения;
- научные источники — исследования в области металлургии и материаловедения.

Такое разнообразие позволяет исследовать общие тематические связи между информационными ресурсами, представляющими различные сегменты металлургического комплекса.

##### 1.7. Предполагаемый результат исследования

На основе собранных веб-страниц будет сформировано множество ключевых терминов для каждого ресурса.

Далее будет проведено попарное сравнение полученных множеств.

Если количество общих терминов между двумя ресурсами достигает установленного порога N, между соответствующими вершинами графа будет создаваться ребро.

Таким образом:

- вершина графа соответствует отдельному веб-ресурсу;
- ребро обозначает тематическую близость двух ресурсов;
- вес ребра определяется количеством общих ключевых терминов;
- толщина ребра отражает степень тематического пересечения.

Полученная визуализация позволит выявить тематически близкие ресурсы, определить группы источников со схожей тематикой и исследовать структуру информационного пространства металлургической промышленности России и ближнего зарубежья.


```python
import os
import sys
import subprocess
from pathlib import Path

PROJECT_DIR = Path.cwd()

env = {
    **os.environ,
    "PYTHONUNBUFFERED": "1",
    "PYTHONIOENCODING": "utf-8",
    "PYTHONUTF8": "1",
}

def run(*args):
    proc = subprocess.Popen(
        [sys.executable, "-u", str(PROJECT_DIR / "run.py"), *args],
        cwd=PROJECT_DIR,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",              # ← родитель читает как UTF-8
        errors="replace",              # ← страховка от одиночных битых байтов
        bufsize=1,
        env=env,
    )

    for line in proc.stdout:
        print(line, end="")

    proc.wait()
    print(f"Код завершения: {proc.returncode}")
```

Получим описание для понимания аргументов

```python
run('-h')
```

Запустим сбор информации и процесс составления графа по вышеописанным ресурсам (metallurgy.txt) c максимальным количеством URL в 30

```python
run('metallurgy.txt', '-mp', '30')
```
